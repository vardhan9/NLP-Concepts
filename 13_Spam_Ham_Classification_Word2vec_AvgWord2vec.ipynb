```python
import gensim
from gensim.models import word2vec, KeyedVectors
```

```python
import pandas as pd
messages = pd.read_csv("/content/SMSSpamCollection.txt", sep='\t', names=["label","message"])
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
```

```python
import nltk
nltk.download('wordnet')
```

```python
import re
corpus = []
for i in range(0,len(messages)):
    review = re.sub('[^a-zA-Z]', '  ',messages['message'][i])
    review = review.lower()
    review= review.split()
    review = [lemmatizer.lemmatize(word) for word in review]
    review = ' '.join(review)
    corpus.append(review)
    print(review)
```

```python
[[i,j,k] for i,j,k in zip(list(map(len,corpus)),corpus, messages['message']) if i<1]
```

```python
# Alternate way of doing
from nltk import sent_tokenize
from gensim.utils import simple_preprocess
```

```python
nltk.download('punkt')
```

```python
words=[]
for sent in corpus:
  sent_token = sent_tokenize(sent)
  for sent in sent_token:
    words.append(simple_preprocess(sent))
```

```python
# each word is represented in a sentence is represented as vector with 100 dimensions
# Avg word2vec is to take average of those words and represent the whole sentence in a vector with 100 dimensions
import numpy
def avg_word2vec(doc):
  return numpy.mean([model.wv[word] for word in doc if word in model.wv.index_to_key], axis=0)
```

```python
X = numpy.empty((len(words), len(avg_word2vec(words[0]))))
for i in range(len(words)):
    X[i] = avg_word2vec(words[i])
```
